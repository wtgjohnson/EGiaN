# SO(N)

Zee presents the sequence of expressions below to justify that the trace of a tensor transforms into itself. In the justification he uses the propery of orthogonal groups that $O^TO=1$. The trace is defined to be $T \equiv \delta^{ij}T^{ij}$. This means that $i=j$ or the term is zero and the terms are summed over all $i$ and $j$. This is equivalent to the trace definition which is the sum of the diagonal elements of a matrix.

$$
T \rightarrow \delta^{ij}T'^{ij} = \delta^{ij}O^{il}O^{jm}T^{lm} = 
(O^T)^{li}\delta^{ij}O^{jm}T^{lm} = \delta^{lm}T^{lm} = T
$$

These terms $T \rightarrow \delta^{ij}T^{ij}$ apply the definition of the trace $T \equiv \delta^{ij}T^{ij}$ and the definition of transformation indicated by the arrow, namely $T^{ij} \rightarrow T'^{ij}$.

These terms $\delta^{ij}T'^{ij} = \delta^{ij}O^{il}O^{jm}T^{lm}$ apply the definition for the transformation of a tensor, namely $T'^{ij} = R^{ik}R^{jl}T^{kl}$ which is equation (1) from chapter I.4 in EGiaN.

These terms $\delta^{ij}O^{il}O^{jm}T^{lm} = (O^T)^{li}\delta^{ij}O^{jm}T^{lm}$ applies transposition to $O^{il}$ twice which does not change the equations. The commutative property of multiplication is also applied. Transposing $O^{il}$ once gives $(O^T)^{il}$. Transposing a second time gives $(O^T)^{li}$. Transposing a matrix twice leaves it unchanged. Zee also moved $\delta^{ij}$ in between the two components from $O$. Since we are working with the individual elements of the matrices as selected by the subscripts, commutativity applies.

The expression $(O^T)^{li}\delta^{ij}O^{jm}$ is zero unless $i=j$. Let $i=j=k$. Dropping the zero terms where $i \ne j$ that must be zero gives

$$
(O^T)^{li}\delta^{ij}O^{jm} = (O^T)^{lk}O^{km} = O^TO = \mathbf{1} = I = \delta^{lm}
$$

Zee doesn't put the 1 in boldface that I've seen. Finally, $\delta^{lm}T^{lm}$ is just the definition of the trace again. So, the trace transforms into itself.

Next, Zee presents these two ways of writing the breakdown of SO(3) into symmetric and antisymmetric subgroups and the trace. I'd like to add some additional explanation concerning these conventions.

$$
3 \otimes 3 = 5 \oplus 1 \oplus 3 \\
\{1\} \otimes \{1\} = \{2\} \oplus \{0\} \oplus [2]
$$

The left-hand side of these equations represent taking the product of two vectors of length 3. In the upper conventions, the length of the vectors is used without the use of any braces or brackets. In the lower convention, the rank of the tensor, that is, the number of indicies of the tensor, is used, and braces are used for symmetric groups and brackets for antisymmetric groups. Vectors have one index and are symmetric. When transposed, a row vector becomes a column vector or the reverse occurs and the elements of the vectors do not change sign. Hence, the left-hand side of the lower convention has two 1's wrapped in {}.

The right-hand side of the upper convention shows the dimensions of each subgroup. There are three subgroups, one each of dimension 5, 1, and 3. No notation is made concerning whether the subgroup is symmetric or antisymmetric. The subgroup of dimension 5 is symmetric. For a $3 \times 3$ two-index tensor, there are 9 elements. Symmetry about tranposision means that $A^{ij}=A^{ji}$. This reduces the number of distinct elements from 9 to 6, the three off-diagonal elements and the diagonal elements. We have recently seen that the trace transforms into itself. So we need to subtract one from this count of distinct elements. Subtracting 1 from 6 yields 5, the dimension of the symmetric subgroup. The subgroup of dimension 1 is the trace's subgroup. The subgroup of dimension 3 is the antisymmetric subgroup. To be antisymmetric $A^{ij}=-A^{ji}$ which implies that the diagonal elements are 0. This leaves three elements that are free to take on distinct values. Hence, the dimnsion of the subgroup is 3.

The right-hand side of the lower convention continues to use the number of indices for the tensor and then wraps the number in braces for a symmetric group and brackets for an antisymmetric group. The product of two vectors to form a tensor produces a tensor with two indices. Hence, there is a symmetric group of two indices {2} and an antisymmetric group of two indices [2]. The trace is of matrix is a scalar and has no indices and since it transforms into itself it is symmetric. It's subgroup is represented by {0}.

When applying these conventions to SO(4), the upper convention changes because the dimensions of the subgroups may get larger, but the lower convention does not change. It is independent of the dimensions of the indices.

$$
4 \otimes 4 = 9 \oplus 1 \oplus 6 \\
\{1\} \otimes \{1\} = \{2\} \oplus \{0\} \oplus [2]
$$

The two vectors now have length of four but still only have one index. The symmetric group of the two-indexed tensor has 9 elements that are free to take on independent values, the off-diagonal values and three more values from the diagonal. The trace group has no indexes can still only take on a single independent value. The antisymmetric group of the two-indexed tensor has 6 elements that are free to take on independent values, the off-diagonal elements only.

Somehow $\varepsilon^{i_1i_2\ldots i_N}O^{i_11}O^{i_22}\ldots O^{i_NN}=1$ is the same thing as $\text{det }O=1$. It was news to me that

$$
\text{det }A = \sum_{i_1,i_2,\ldots,i_N} \varepsilon^{i_1 i_2 \ldots i_N} A_{i_11}A_{i_22}\ldots A_{i_NN}
$$

I think that means that for a $3\times 3$ matrix we have:

$$
\text{det }A = \sum_{i_1,i_2,i_3} \varepsilon^{i_1 i_2 i_3} A_{i_1 1}A_{i_2 2}A_{i_3 3} = \\
\varepsilon^{111}A_{11}A_{12}A_{13} + \varepsilon^{112}A_{11}A_{12}A_{23} + \varepsilon^{113}A_{11}A_{12}A_{33} + 
\varepsilon^{121}A_{11}A_{22}A_{13} + \varepsilon^{122}A_{11}A_{22}A_{23} + \varepsilon^{123}A_{11}A_{22}A_{33} +
\varepsilon^{131}A_{11}A_{32}A_{13} + \varepsilon^{132}A_{11}A_{32}A_{23} + \varepsilon^{133}A_{11}A_{32}A_{33} + 
\varepsilon^{211}A_{21}A_{12}A_{13} + \varepsilon^{212}A_{21}A_{12}A_{23} + \varepsilon^{213}A_{21}A_{12}A_{33} +
\varepsilon^{221}A_{21}A_{22}A_{13} + \varepsilon^{222}A_{21}A_{22}A_{23} + \varepsilon^{223}A_{21}A_{22}A_{33} + 
\varepsilon^{231}A_{21}A_{32}A_{13} + \varepsilon^{232}A_{21}A_{32}A_{23} + \varepsilon^{233}A_{21}A_{32}A_{33} + 
\varepsilon^{311}A_{31}A_{12}A_{13} + \varepsilon^{312}A_{31}A_{12}A_{23} + \varepsilon^{313}A_{31}A_{12}A_{33} + 
\varepsilon^{321}A_{31}A_{22}A_{13} + \varepsilon^{322}A_{31}A_{22}A_{23} + \varepsilon^{323}A_{31}A_{22}A_{33} + 
\varepsilon^{331}A_{31}A_{32}A_{13} + \varepsilon^{332}A_{31}A_{32}A_{23} + \varepsilon^{333}A_{31}A_{32}A_{33}
$$

Every element of $\varepsilon$ where an index is repeated is 0 because $\varepsilon$ is defined to be antisymmetric under the exchange of indices. So, if two indices are identical and the element must equal its own negative, then it must be zero. Dropping the zero terms from above leaves:

$$
\text{det }A = \varepsilon^{123}A_{11}A_{22}A_{33} + \varepsilon^{132}A_{11}A_{32}A_{23} + \varepsilon^{213}A_{21}A_{12}A_{33} + \varepsilon^{231}A_{21}A_{32}A_{13} + \varepsilon^{312}A_{31}A_{12}A_{23} + \varepsilon^{321}A_{31}A_{22}A_{13} 
$$

And since $\varepsilon$ is antisymmetric under the interchange of any two indices and $\varepsilon^{123}=1$ we have:

$$
\varepsilon^{123} = 1 \\
\varepsilon^{132} = -1 \\
\varepsilon^{213} = -1 \\
\varepsilon^{231} = 1 \\
\varepsilon^{312} = 1 \\
\varepsilon^{321} = -1 \\
$$

which gives

$$
\text{det }A = A_{11}A_{22}A_{33} - A_{11}A_{32}A_{23} - A_{21}A_{12}A_{33} + A_{21}A_{32}A_{13} + A_{31}A_{12}A_{23} - A_{31}A_{22}A_{13}
$$

The formula that I learned for the determinant of a $3 \times 3$ matrix was:

$$
\text{det }A = A_{11}(A_{22}A_{33}-A_{23}A_{32}) - A_{12}(A_{21}A_{33}-A_{23}A_{31}) + A_{13}(A_{21}A_{32}-A_{22}A_{31}) = \\
A_{11}A_{22}A_{33} - A_{11}A_{23}A_{32} - A_{12}A_{21}A_{33} + A_{12}A_{23}A_{31} + A_{13}A_{21}A_{32} - A_{13}A_{22}A_{31}
$$

So, for a $3 \times 3$ matrix, they are the same. I'll be damned. I'm willing to accept that this trick works for square matrices of any size.
